```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd

url = "https://raw.githubusercontent.com/SyedAmmarSharif/Flyrank-ML-Internship-Work/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Total content pages:", len(df))

declining_pages = (df["trend_direction"] == "down").sum()

print("Declining pages:", declining_pages)
print("Declining percentage:", round((declining_pages / len(df)) * 100, 2), "%")

visible_pages = (df["impressions_90d"] >= 500).sum()

print("Pages with at least 500 impressions:", visible_pages)
print("Percentage of visible pages:", round((visible_pages / len(df)) * 100, 2), "%")
```
